# Managing Multiple Caches

Fleche makes it easy to switch between different caches or move data between them using `fleche.cache` and the `transfer()` method.

```python
from fleche import fleche, cache
from fleche.caches import BaseCache

# Create two separate memory caches
cache_A = BaseCache.from_config({"template": "memory"})
cache_B = BaseCache.from_config({"template": "memory"})
```

## Switching Caches

You can use the `cache` context manager to temporarily switch the active cache. Any functions decorated with `@fleche` called inside this block will use the specified cache.

```python
@fleche
def compute_value(x):
    return x * 10

# Run and store inside cache_A
with cache(cache_A):
    compute_value(5)
    print("Is 5 in cache_A?", compute_value.fleche.contains(5))

with cache(cache_B):
    print("Is 5 in cache_B?", compute_value.fleche.contains(5))
```

## Transferring Data Between Caches

The `transfer()` method allows you to copy or move cache entries from one cache to another. Passing `pop=True` will evict the transferred items from the source cache, essentially moving them instead of copying.

```python
# Transfer all entries from cache_A to cache_B and remove them from cache_A
cache_A.transfer(cache_B, pop=True)

print("After transferring with pop=True:")
with cache(cache_A):
    print("Is 5 in cache_A?", compute_value.fleche.contains(5))

with cache(cache_B):
    print("Is 5 in cache_B?", compute_value.fleche.contains(5))
```
